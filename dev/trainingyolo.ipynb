### Plot images

```python
# Load your trained model
model1 = YOLO('/home/krschap/Jupyter/manu/drone-imagery/building-footprint/dev/test_files/first_run/weights/best.pt')  
image_paths = list_imgs[:10] 

results = model1.predict(
    source=image_paths,  # can pass a list directly
    conf=0.25,           # confidence threshold
    save=True,           # save results to runs/detect/predict
    save_txt=True,       # save labels as txt files
    save_conf=True,      # save confidences in labels
)

n_images = len(list_imgs[:10])


# Plot results using matplotlib
n_images = len(results)
cols = 3  # number of columns in plot
rows = (n_images + cols - 1) // cols  # calculate rows needed

fig, axes = plt.subplots(rows, cols, figsize=(15, 5 * rows))
axes = axes.flatten() if n_images > 1 else [axes]

for idx, result in enumerate(results):
    # Get the image with predictions drawn on it
    img_with_predictions = result.plot()  # returns numpy array with boxes drawn
    
    # Plot
    axes[idx].imshow(img_with_predictions)
    axes[idx].axis('off')
    axes[idx].set_title(f'Image {idx + 1}\nDetections: {len(result.boxes)}')
    
    # Optional: print detection details
    print(f"\nImage {idx + 1}: {os.path.basename(image_paths[idx])}")
    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        class_name = result.names[class_id]
        print(f"  - {class_name}: {confidence:.2f}")

# Hide empty subplots if any
for idx in range(n_images, len(axes)):
    axes[idx].axis('off')

plt.tight_layout()
plt.show()
```
